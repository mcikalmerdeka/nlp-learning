```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter
from langchain_core.documents import Document
from pprint import pprint

# Load the markdown file
doc_path = "ehr_database_docs.md"
loader = TextLoader(doc_path)
docs = loader.load()  # Returns list with 1 Document

# Define headers to split on (all levels)
headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3"),
    ("####", "Header 4"),
]

# Split by headers, preserving hierarchy in metadata
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on,
    strip_headers=False  # Keep header text in content
)
splits = splitter.split_text(docs[0].page_content)

# Convert to Documents with metadata
chunked_docs = [Document(page_content=s.page_content, metadata=s.metadata) for s in splits]
```
